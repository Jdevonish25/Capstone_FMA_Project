```python
# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ============================================================
# 2. CREATE FINAL RESULTS TABLE
# ============================================================

final_results = pd.DataFrame([
    {
        "Scenario": "8-Genre Validated System",
        "Structured Model": "Random Forest",
        "Structured Accuracy": 0.4053333333333333,
        "Audio Model": "Improved Audio CNN",
        "Audio Accuracy": 0.4533333480358124,
        "Hybrid Accuracy": 0.4613333333333333,
        "Notes": "Main validated proof-of-concept pipeline"
    },
    {
        "Scenario": "Expanded 8-Genre Experiment",
        "Structured Model": "Random Forest",
        "Structured Accuracy": 0.41333333333333333,
        "Audio Model": "Improved Audio CNN",
        "Audio Accuracy": 0.41333332657814026,
        "Hybrid Accuracy": 0.44266666666666665,
        "Notes": "More training data alone did not improve final hybrid performance"
    },
    {
        "Scenario": "10-Genre Broader Extension",
        "Structured Model": "SVM",
        "Structured Accuracy": 0.43177966101694915,
        "Audio Model": "Improved Audio CNN",
        "Audio Accuracy": 0.31409168243408203,
        "Hybrid Accuracy": 0.47156196943972833,
        "Notes": "Broader label space with full validation/test coverage"
    }
])

print("Final Results Summary:")
display(final_results)
```

```python
# ============================================================
# 3. SAVE FINAL RESULTS TABLE
# ============================================================

os.makedirs("../data/processed", exist_ok=True)

final_results.to_csv(
    "../data/processed/final_results_summary.csv",
    index=False
)

print("Saved final results summary to ../data/processed/final_results_summary.csv")
```

```python
# ============================================================
# 4. PLOT STRUCTURED VS AUDIO VS HYBRID BY SCENARIO
# ============================================================

plot_df = final_results.set_index("Scenario")[
    ["Structured Accuracy", "Audio Accuracy", "Hybrid Accuracy"]
]

ax = plot_df.plot(kind="bar", figsize=(11, 6))
ax.set_ylabel("Accuracy")
ax.set_title("Structured vs Audio vs Hybrid Accuracy by Scenario")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 5. PLOT HYBRID ACCURACY ONLY
# ============================================================

plt.figure(figsize=(9, 5))
plt.bar(final_results["Scenario"], final_results["Hybrid Accuracy"])
plt.ylabel("Hybrid Accuracy")
plt.title("Final Hybrid Accuracy by Scenario")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 6. CREATE FINAL INTERPRETATION TABLE
# ============================================================

interpretation_df = pd.DataFrame([
    {
        "Finding": "Hybrid fusion was consistently the best approach",
        "Meaning": "Across both the 8-genre and 10-genre systems, combining structured and audio branches gave the strongest final performance."
    },
    {
        "Finding": "The expanded 8-genre experiment did not improve the final model",
        "Meaning": "Increasing training data alone was not enough to improve the hybrid pipeline under the current setup."
    },
    {
        "Finding": "The 10-genre extension remained viable",
        "Meaning": "Even on the broader and harder 10-genre task, the hybrid model still outperformed the individual structured and audio branches."
    },
    {
        "Finding": "The deployed pipelines worked end to end",
        "Meaning": "Final inference notebooks successfully accepted audio input, extracted features, generated branch predictions, fused them, and returned final genre outputs."
    }
])

print("Final Interpretation Table:")
display(interpretation_df)
```

```python
# ============================================================
# 7. SAVE FINAL INTERPRETATION TABLE
# ============================================================

interpretation_df.to_csv(
    "../data/processed/final_interpretation_summary.csv",
    index=False
)

print("Saved interpretation summary to ../data/processed/final_interpretation_summary.csv")
```
